```python
from google import genai
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, AIMessage

class ChatBot:
    def __init__(self, api_key, model_id="gemini-3.1-flash-lite-preview"):
        self.api_key = api_key
        self.model_id = model_id

        # 1. Setup Native GenAI SDK Session
        self.client = genai.Client(api_key=api_key)
        self.genai_chat = self.client.chats.create(model=self.model_id)

        # 2. Setup LangChain Session
        self.lc_history = []
        self.llm = ChatGoogleGenerativeAI(model=self.model_id, api_key=api_key)

    def ask_genai(self, prompt):
        """Uses the native Google GenAI SDK (Internal State)"""
        response = self.genai_chat.send_message(prompt)
        return response.text

    def ask_langchain(self, prompt):
        """Uses LangChain Logic (Manual History Management)"""
        # Append User Message
        self.lc_history.append(HumanMessage(content=prompt))

        # Invoke Model
        response = self.llm.invoke(self.lc_history)

        # Append AI Message
        self.lc_history.append(AIMessage(content=response.content))
        return response.content[0]["text"]

#  Execution
api_key = google_api_key

# Initialize our combined developer tool
bot = ChatBot(api_key=api_key)
```

```python
# Scenario 1: Using Native GenAI SDK
print("Native GenAI SDK Model")
print(f"User: My name is Shiva.")
print(f"AI: {bot.ask_genai('My name is Shiva')}")
print(f"User: What is my name?")
print(f"AI: {bot.ask_genai('1')}")
```

```python
# Scenario 2: Using LangChain
print("\n LangChain Model ")
print(f"User: I am a student")
print(f"AI: {bot.ask_langchain('I am a student')}")
print(f"User: What is my profession?")
print(f"AI: {bot.ask_langchain('What is my profession?')}")
```

```python
def calculator():
  print(" Simple CLI Calculator")
  a = float(input("Enter first number: "))
  x = input("Enter operator (+, -, *, /): ")
  b = float(input("Enter second number: "))

  if x == '+':
      result = a + b
  elif x == '-':
      result = a - b
  elif x == '*':
      result = a * b
  elif x == '/':
      if b != 0:
          result = a / b
      else:
          return "Error: Division by zero is undefined."
  else:
      return "Error: Invalid operator."

  return f"Result: {a} {x} {b} = {result}"

print(calculator())
```

```python
def evaluate_student():
    print("--- Student Evaluation System ---")

    # Get user input for age and marks
    try:
        age = int(input("Enter the student's age: "))
        marks = float(input("Enter the student's marks (0-100): "))
    except ValueError:
        print("Invalid input! Please enter numbers only.")
        return

    print("\n--- Evaluation Results ---")

    # 1. Voting and License Check
    if age >= 18:
        print("Legal Status: Eligible for Voting and Driving License.")
    else:
        print("Legal Status: Minor. Not eligible for Voting or License yet.")

    # 2. Marks and Grading Check
    if marks >= 90:
        print("Grade: A+ (Excellent)")
    elif marks >= 75:
        print("Grade: A (Good)")
    elif marks >= 50:
        print("Grade: B (Average)")
    else:
        print("Grade: Fail")

    # 3. Course Eligibility Check (Nested If-Else)
    if age >= 18:
        if marks >= 60:
            print("Course Allocation: Enrolled in Advanced IT Course.")
        else:
            print("Course Allocation: Enrolled in Basic Foundation Course.")
    else:
        if marks >= 75:
            print("Course Allocation: Enrolled in Junior Honors Program.")
        else:
            print("Course Allocation: Enrolled in Standard High School Curriculum.")

# Run the program
evaluate_student()
```

```python
def calculate_student_grade():
    print("--- Student Grade Calculator ---")
    student_name = input("Enter student's name: ")
    try:
        num_subjects = int(input("Enter the number of subjects: "))
        if num_subjects <= 0:
            print("Number of subjects must be greater than 0.")
            return
    except ValueError:
        print("Invalid input! Please enter a whole number.")
        return
    total_marks = 0
    for i in range(1, num_subjects + 1):
        while True:
            try:
                mark = float(input(f"Enter marks for subject {i} (0-100): "))
                if 0 <= mark <= 100:
                    total_marks += mark
                    break
                else:
                    print("Please enter a value between 0 and 100.")
            except ValueError:
                print("Invalid input! Please enter a number.")

    average_marks = total_marks / num_subjects


    if average_marks >= 90:
        grade = "A+"
    elif average_marks >= 80:
        grade = "A"
    elif average_marks >= 70:
        grade = "B"
    elif average_marks >= 60:
        grade = "C"
    elif average_marks >= 50:
        grade = "D"
    else:
        grade = "Fail"
    print("\n=========================")
    print("      REPORT CARD        ")
    print("=========================")
    print(f"Student Name  : {student_name}")
    print(f"Total Marks   : {total_marks} / {num_subjects * 100}")
    print(f"Average Marks : {average_marks:.2f}")
    print(f"Final Grade   : {grade}")
    print("=========================")
calculate_student_grade()
```

```python
total_sum = 0
print("Even/Odd Checker and Adder")
print("Enter 0 at any time to stop the program.\n")
while True:
    try:
        num = int(input("Enter a number: "))
        if num == 0:
            print("Stopping the loop...")
            break
        if num % 2 == 0:
            print(f"{num} is Even.")
        else:
            print(f"{num} is Odd.")
        total_sum += num
    except ValueError:
        print("Invalid input! Please enter a valid integer.")
print(f"The total sum of all numbers is: {total_sum}")
```

```python
sentence = input("Enter a sentence: ")
total_chars = len(sentence)
total_words = len(sentence.split())

upper_case_count = 0
lower_case_count = 0

for char in sentence:
    if char.isupper():
        upper_case_count += 1
    elif char.islower():
        lower_case_count += 1

words_to_replace = {
    "hi": "hello",
    "bad": "good",
    "sad": "happy",
    "hate": "love"
}

modified_sentence = sentence

for old_word, new_word in words_to_replace.items():
    modified_sentence = modified_sentence.replace(old_word, new_word)
    modified_sentence = modified_sentence.replace(old_word.capitalize(), new_word.capitalize())

x = sentence.upper()
y = sentence.lower()

print("\nResults")
print(f"Original Sentence: {sentence}")
print(f"Total Characters: {total_chars}")
print(f"Total Words: {total_words}")
print(f"Uppercase Characters: {upper_case_count}")
print(f"Lowercase Characters: {lower_case_count}")
print(f"Modified Sentence: {modified_sentence}")
print(f"Uppercase Sentence: {x}")
print(f"Lowercase Sentence: {y}")
```

```python
marks_list = []

print("Enter student marks one by one.")
print("Enter 0 to stop and see the results.\n")

while True:
    try:
        mark = float(input("Enter student mark: "))
        if mark == 0:
            break

        marks_list.append(mark)
    except ValueError:
        print("Invalid input! Please enter a valid number.")

pass_count = 0
fail_count = 0
total_marks = 0
pass_total = 0
fail_total = 0

for mark in marks_list:
    total_marks += mark

    if mark >= 60:
        pass_count += 1
        pass_total += mark
    else:
        fail_count += 1
        fail_total += mark

total_students = len(marks_list)

average_all = total_marks / total_students if total_students > 0 else 0
average_pass = pass_total / pass_count if pass_count > 0 else 0
average_fail = fail_total / fail_count if fail_count > 0 else 0

print("       FINAL RESULTS")

if total_students == 0:
    print("No marks were entered.")
else:
    print(f"Total students:       {total_students}")
    print(f"Total marks:          {total_marks}")
    print(f"Overall average:      {average_all:.2f}")
    print("-" * 30)
    print(f"Students passed:      {pass_count}")
    print(f"Average passing mark: {average_pass:.2f}")
    print("-" * 30)
    print(f"Students failed:      {fail_count}")
    print(f"Average failing mark: {average_fail:.2f}")
```

```python
def process_student_name():
    print("--- Student Name Processing ---")
    name = input("Enter student name: ")

    vowels = "aeiouAEIOU"
    vowel_count = 0
    for char in name:
        if char in vowels:
            vowel_count += 1

    name_length = len(name)

    if name_length % 2 == 0:
        length_status = "Even"
        converted_name = name.upper()
    else:
        length_status = "Odd"
        converted_name = name.lower()

    # 6. Print the results
    print("\n--- Results ---")
    print(f"Original Name  : {name}")
    print(f"Total Vowels   : {vowel_count}")
    print(f"Name Length    : {name_length} characters")
    print(f"Length is      : {length_status}")
    print(f"Converted Name : {converted_name}")

# Run the program
if __name__ == "__main__":
    process_student_name()
```
